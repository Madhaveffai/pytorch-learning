nn.module

```python
import torch
import torch.nn as nn

class LinearModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear = nn.Linear(1,1)

  def forward(self, x):
    return self.linear(x)

model = LinearModel()
print(model)
```

Inspect Parameters

```python
for name, param in model.named_parameters():
  print(name, param.shape, param.data)
```

Data

```python
X = torch.linspace(0, 1, 20).unsqueeze(1)
y = 3 * X + 2 + 0.1 * torch.randn(20,1)
```

Training Loop

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)
loss_fn = nn.MSELoss()

for epoch in range(100):
  y_pred = model(X)
  loss = loss_fn(y_pred, y)

  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  if epoch % 10 == 0:
    print(f"Epoch {epoch} | loss: {loss.item():.4f}")
```

```python
print(model.linear.weight.item())
print(model.linear.bias.item())
```
