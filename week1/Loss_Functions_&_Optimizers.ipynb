MSELoss - regression

```python
import torch
import torch.nn as nn

loss_fn = nn.MSELoss()

y_pred = torch.tensor([2.5, 3.1, 4.0])
y_true = torch.tensor([2.0, 3.0, 4.5])

print(loss_fn(y_pred, y_true))
```

CrossEntropyLoss - Classification

```python
loss_fn = nn.CrossEntropyLoss()

y_pred = torch.tensor([[1.0, 2.0, 0.5, 0.1],
                       [0.2, 0.1, 3.0, 0.4],
                       [1.5, 0.3, 0.2, 2.1]])
y_true = torch.tensor([1, 2, 3])

print(loss_fn(y_pred, y_true))
```

Optimizer

```python
X = torch.linspace(0, 1, 20).unsqueeze(1)
y = 3 * X + 2 + 0.1 * torch.randn(20, 1)

for opt_name, opt_class in [("SGD", torch.optim.SGD), ("Adam", torch.optim.Adam)]:
  model = nn.Sequential(nn.Linear(1,1))
  optimizer = opt_class(model.parameters(), lr=0.001)
  loss_fn = nn.MSELoss()

  for epoch in range(1000):
    loss = loss_fn(model(X), y)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

  print(f"{opt_name} final loss: {loss.item():.4f}")
```
