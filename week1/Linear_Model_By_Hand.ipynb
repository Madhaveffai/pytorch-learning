Create fake data

```python
import torch

x = torch.linspace(0, 1, 20).unsqueeze(1)
y = 3 * x + 2 + 0.2 * torch.randn(20,1)
```

Initialize Weights

```python
W = torch.randn(1, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

Training Loop

```python
lr = 0.1

for epoch in range(100):
  y_pred = x @ W + b
  loss = ((y_pred - y)  ** 2).mean()
  loss.backward()
  with torch.no_grad():
    W -= lr * W.grad
    b -= lr * b.grad

  W.grad.zero_()
  b.grad.zero_()

  if epoch % 10 == 0:
    print(f"Epoch {epoch} | Loss: {loss.item():.4f}")
```

Check What the model learned

```python
print(f"W = {W.item():.2f}, b = {b.item():.2f}")
```
