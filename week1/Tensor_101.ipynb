Creating Tensors

```python
import torch

a = torch.tensor([1.0, 2.0, 3.0])
b = torch.zeros(3,4)
c = torch.ones(2,3)
d = torch.rand(3,3)

print(a.shape)
print(b.shape)
print(d.dtype)
```

Reshaping and Indexing

```python
x = torch.arange(12, dtype=torch.float32) #[0,1,2,3,4....11]
x_2d = x.reshape(3,4) #3 rows and 4 cols
x_2d_T = x_2d.T #transpose

print(x_2d)
print(x_2d[0]) # first row
print(x_2d[:, 1]) # second column
print( x_2d[1, 2])
```
