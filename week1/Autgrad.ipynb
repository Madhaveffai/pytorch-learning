Basic Gradient

```python
import torch

x = torch.tensor(3.0, requires_grad=True)
y = x ** 2

y.backward()
print(x.grad)
```

A small computation graph

```python
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(5.0, requires_grad=True)

y = w * x + 3
y.backward()

print(x.grad)
print(w.grad)
```

Gradients accumulate, zeroing them

```python
x = torch.tensor(3.0, requires_grad=True)

for i in range(3):
    y = x ** 2
    y.backward()
    print(x.grad)

# Reset before each pass
x.grad.zero_()
```

Stop tracking with no_grad

```python
x = torch.tensor(3.0, requires_grad=True)

with torch.no_grad():
    y = x ** 2    # no graph built, no gradient tracked
    print(y)      # tensor(9.) just a value
```

Define x = 4.0 with grad tracking. Compute y = 3x³. Call backward. What should x.grad be? Verify it.

```python
x = torch.tensor(4.0, requires_grad=True)

y = 3 * x ** 3
y.backward()
print(x.grad)
```
