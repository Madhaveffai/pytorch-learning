Save a Model

```python
import torch
import torch.nn as nn

class MLP(nn.Module):
  def __init__(self):
    super().__init__()
    self.net = nn.Sequential(
        nn.Linear(1, 32), nn.ReLU(),
        nn.Linear(32, 1)
    )
  def forward(self, x):
    return self.net(x)

model = MLP()

optimizer = torch.optim.Adam(model.parameters(), lr = 0.01)
loss_fn = nn.MSELoss()
X = torch.linspace(-3, 3, 100).unsqueeze(1)
y = torch.sin(X) + 0.1 * torch.randn(100, 1)

for epoch in range(100):
  loss = loss_fn(model(X), y)
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

print(f"Trained loss: {loss.item():.4f}")

torch.save(model.state_dict(), "model.pth")
print("Model Saved")
```

Load it back

```python
model2 = MLP()

model2.load_state_dict(torch.load("model.pth"))
model2.eval()

with torch.no_grad():
  pred = model2(torch.tensor([[1.0]]))
  print(f"Prediction for x=1: {pred.item():.4f}")
```

Save a checkpoint

```python
torch.save({
    'epoch': 100,
    'model_state_dict': model.state_dict(),
    'optimizer_state_dict': optimizer.state_dict(),
    'loss': loss.item()
}, "checkpoint.pth")

checkpoint = torch.load("checkpoint.pth")
model3 = MLP()
optimizer3 = torch.optim.Adam(model3.parameters(), lr=0.01)

model3.load_state_dict(checkpoint['model_state_dict'])
optimizer3.load_state_dict(checkpoint['optimizer_state_dict'])
resume_epoch = checkpoint['epoch']

print(f"Resumed from epoch {resume_epoch}")
```

```python
model2 = MLP()

model2.load_state_dict(torch.load("model.pth"))
model2.eval()

with torch.no_grad():
  pred0 = model2(torch.tensor([[0.0]]))
  pred1 = model2(torch.tensor([[1.5]]))
  print(f"Prediction for x = 0.0: {pred0.item():.4f}")
  print(f"Prediction for x = 1.5: {pred1.item():.4f}")
```
