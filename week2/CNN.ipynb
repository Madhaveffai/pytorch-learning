Key Layers

```python
import torch
import torch.nn as nn

conv = nn.Conv2d(1, 8, kernel_size=3, padding=1)

pool = nn.MaxPool2d(2)

x = torch.randn(4, 1, 28, 28)
print(x.shape)

x = conv(x)
print(x.shape)

x = pool(x)
print(x.shape)
```

Shape Formula

out = (in + 2*padding - kernel_size) / stride + 1\
With in=28, padding=1, kernel=3, stride=1:

```python
out = (28 + 2 - 3)/1 + 1
```

Build a small CNN

```python
class SmallCNN(nn.Module):
  def __init__(self, num_classes=10):
    super().__init__()
    self.features = nn.Sequential(
        nn.Conv2d(1, 8, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.MaxPool2d(2),
        nn.Conv2d(8, 16, kernel_size=3, padding=1),
        nn.ReLU(),
        nn.MaxPool2d(2),
    )
    self.classifier = nn.Sequential(
        nn.Linear(16 * 7 * 7, 64),
        nn.ReLU(),
        nn.Linear(64, num_classes)
    )

  def forward(self, x):
    x = self.features(x)
    x = x.view(x.size(0), -1)
    x = self.classifier(x)
    return x
model = SmallCNN()
print(model)
```

Sanity Check

```python
x = torch.randn(4, 1, 28, 28)
out = model(x)
print(out.shape)
```
