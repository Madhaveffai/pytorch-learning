Basic MathOps

```python
import torch

a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

print(a + b)
print(a * b)
print(a ** 2)
print(torch.sqrt(a))

A = torch.rand(3,4)
B = torch.rand(4,2)
print(A@B)
print(A.shape, B.shape)
```

Broadcasting

```python
x = torch.ones(3,4)
v = torch.tensor([1.0, 2.0, 3.0, 4.0])

print(x + v)
```

In-place vs out-of-place

```python
x = torch.tensor([1.0, 2.0, 3.0])

y = x + 1
x.add_(1)

print(x)
```

Reductions

```python
x = torch.rand(3,4)

print(x.sum())
print(x.sum(dim=0)) #sum along rows
print(x.sum(dim=1)) #sum along cols
print(x.mean())
print(x.max())
print(x.argmax())
```

Create two random tensors of shape (4, 3) and (3, 5). Matrix multiply them. Then sum the result along dim=0 and print the shape.

```python
a = torch.rand(4,3)
b = torch.rand(3,5)

y = a@b

print(y.sum(dim=0))
print(y.sum(dim=0).shape)
```
